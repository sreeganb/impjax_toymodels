Interfacial interchain residue pairs list from input pdb

```python
from Bio.PDB import PDBParser

# 1. Read the PDB file
parser = PDBParser(QUIET=True)
structure = parser.get_structure("protein", "./pdb/kcoil_ecoil.pdb")
model = structure[0]  # Select the first model

# 2. Extract chains A and B
chain_a = model["A"]
chain_b = model["B"]

# Filter out hetero-residues (like water or ligands)
res_a = [r for r in chain_a.get_residues() if r.id[0] == " "]
res_b = [r for r in chain_b.get_residues() if r.id[0] == " "]

# 3. Compute distances between residue pairs from different chains (A and B)
distances = []
for ra in res_a:
  if "CA" not in ra:
    continue
  for rb in res_b:
    if "CA" not in rb:
      continue

    # Calculate distance between alpha carbons (CA)
    dist = ra["CA"] - rb["CA"]
    distances.append((dist, ra, rb))

# 4. Sort distances in ascending order
distances.sort(key=lambda x: x[0])

# 5. Print the top 25 closest residue pairs
print(f"Top 25 closest residue pairs between Chain A and Chain B:")
for i, (dist, ra, rb) in enumerate(distances[:25], 1):
  print(
      f"{i:2d}. Chain A: {ra.resname} {ra.id[1]} <---> "
      f"Chain B: {rb.resname} {rb.id[1]} | Distance: {dist:.2f} Å"
  )
```
